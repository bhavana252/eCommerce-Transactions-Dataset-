```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Load the datasets
customers_df = pd.read_csv(r'C:\Users\WIN-11\Desktop\Zeotap Internship\data\Customers.csv')
products_df = pd.read_csv(r'C:\Users\WIN-11\Desktop\Zeotap Internship\data\Products.csv')
transactions_df = pd.read_csv(r'C:\Users\WIN-11\Desktop\Zeotap Internship\data\Transactions.csv')

customers_df['SignupDate'] = pd.to_datetime(customers_df['SignupDate'])
transactions_df['TransactionDate'] = pd.to_datetime(transactions_df['TransactionDate'])
```

```python
# Check for missing values in each dataset
print("\nMissing values in Customers DataFrame:")
print(customers_df.isnull().sum())

print("\nMissing values in Products DataFrame:")
print(products_df.isnull().sum())

print("\nMissing values in Transactions DataFrame:")
print(transactions_df.isnull().sum())
```

```python
customers_df.dtypes
```

```python
# Summary statistics for numeric columns
print("\nCustomers Summary Statistics:")
print(customers_df.describe())

print("\nProducts Summary Statistics:")
print(products_df.describe())

print("\nTransactions Summary Statistics:")
print(transactions_df.describe())
# There is no duplicates in data as cust id and names are unique
```

```python
# Select numerical columns for the box plot
numerical_columns = ['Quantity', 'TotalValue', 'Price']

# Create box plots for Quantity, TotalValue, and Price
plt.figure(figsize=(14, 8))

# Plot Quantity, TotalValue, and Price side by side
plt.subplot(2, 2, 1)
sns.boxplot(data=transactions_df, x='Quantity')
plt.title('Box Plot for Quantity')

plt.subplot(2, 2, 2)
sns.boxplot(data=transactions_df, x='TotalValue')
plt.title('Box Plot for TotalValue')

plt.subplot(2, 2, 3)
sns.boxplot(data=transactions_df, x='Price')
plt.title('Box Plot for Price')


# Show the plots
plt.tight_layout()
plt.show()
```

```python
# Get the minimum and maximum transaction dates
min_date = transactions_df['TransactionDate'].min()
max_date = transactions_df['TransactionDate'].max()

# Display the result
print("Earliest Transaction Date:", min_date)
print("Latest Transaction Date  :", max_date)
```

```python
# Plot distribution of product prices
plt.figure(figsize=(8, 6))
sns.histplot(products_df['Price'], bins=20, kde=True)
plt.title('Distribution of Product Prices')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency')
plt.show()

```

```python

# Plot the number of transactions over time
plt.figure(figsize=(10, 6))
transactions_df['Month'] = transactions_df['TransactionDate'].dt.to_period('M')
transactions_per_month = transactions_df.groupby('Month').size()
transactions_per_month.plot(kind='line', color='b')
plt.title('Number of Transactions Over Time')
plt.xlabel('Month')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45)
plt.show()
# Highest transactions happend in Jan, based on given data
```

```python

# Transaction total by product category
product_category_sales = transactions_df.merge(products_df[['ProductID', 'Category']], on='ProductID')
category_sales = product_category_sales.groupby('Category')['TotalValue'].sum().sort_values(ascending=False)
plt.figure(figsize=(10, 6))
category_sales.plot(kind='bar', color='c')
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (USD)')
plt.xticks(rotation=45)
plt.show()
```

```python

# Customer region distribution
region_counts = customers_df['Region'].value_counts()
plt.figure(figsize=(8, 6))
sns.barplot(x=region_counts.index, y=region_counts.values, palette='Set2')
plt.title('Number of Customers by Region')
plt.xlabel('Region')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()

# insights cust are almost distributed evenly but most of them are from SA
```

```python
# Top 10 customers based on total spending -- insight 3
customer_spending = transactions_df.groupby('CustomerID')['TotalValue'].sum().sort_values(ascending=False).head(10)
top_customers = customers_df[customers_df['CustomerID'].isin(customer_spending.index)]
top_customers['TotalSpending'] = top_customers['CustomerID'].map(customer_spending)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_customers['CustomerName'], y=top_customers['TotalSpending'], palette='viridis')
plt.title('Top 10 Customers by Total Spending')
plt.xlabel('Customer Name')
plt.ylabel('Total Spending (USD)')
plt.xticks(rotation=90)
plt.show()
```

```python

# Product performance based on quantity sold
product_sales = transactions_df.groupby('ProductID')['Quantity'].sum().sort_values(ascending=False).head(5)
top_selling_products = products_df[products_df['ProductID'].isin(product_sales.index)]
top_selling_products['TotalQuantitySold'] = top_selling_products['ProductID'].map(product_sales)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_selling_products['ProductName'], y=top_selling_products['TotalQuantitySold'], palette='magma')
plt.title('Top 5 Best Selling Products by Quantity Sold')
plt.xlabel('Product Name')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=90)
plt.show()
```

```python
# top products --------- insight 1
# Product performance based on quantity sold
product_sales = transactions_df.groupby('ProductID')['Price'].sum().sort_values( ascending=False).head(5)

top_selling_products = products_df[products_df['ProductID'].isin(product_sales.index)]
top_selling_products['TotalRevenue'] = top_selling_products['ProductID'].map(product_sales)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_selling_products['ProductName'], y=top_selling_products['Price'], palette='magma')
plt.title('Top 5 Best Selling Products by Revenue')
plt.xlabel('Product Name')
plt.ylabel('Revenue')
plt.xticks(rotation=90)
plt.show()
```

```python

# Analyze signup date vs. first purchase behavior
customers_df['FirstTransactionDate'] = customers_df['CustomerID'].map(
    transactions_df.groupby('CustomerID')['TransactionDate'].min())
customers_df['DaysToFirstPurchase'] = (customers_df['FirstTransactionDate'] - customers_df['SignupDate']).dt.days

plt.figure(figsize=(8, 6))
sns.histplot(customers_df['DaysToFirstPurchase'], bins=30, kde=True)
plt.title('Days to First Purchase After Signup')
plt.xlabel('Days to First Purchase')
plt.ylabel('Frequency')
plt.show()

```
